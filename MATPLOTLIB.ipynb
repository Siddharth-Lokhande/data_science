```python
%matplotlib inline
from matplotlib import pyplot as plt
```

plotting array

```python
plt.plot([1,2,3,4],[4,16,9,5])
plt.show()
```

```python
plt.plot([1,2,3,4],[1,4,9,16])
plt.title("square number")
plt.xlabel("number")
plt.ylabel("square")
plt.show()
```

```python
x=[1,5,4]
y=[5,1,3]
x1=[2,6,2]
y1=[6,2,2]
plt.plot(x,y,'g',label='line one',linewidth=5)
plt.plot(x1,y1,label='line two')
plt.legend()
plt.show()
```
